# Observation from Task 1
There are 50 rows and 6 columns in the data. Each rows contains the employee specifics of a certain employee in the company.

The features in the data set are:
  1. The id of the employees
  2. Their respective groups and age
  3. Their healthy_eating and active_lifestyle
  4. Their salary

# Task 2 - Visualize the distributions of ratings and compensations
We will now Visualize the distributions of employee ratings and compensations

We will create the following plots:

*   A scatter plot of employee age and employee salary, healthy_eating and active_lifestyle, health_eating and salary
*   A count plot of groups
*   A histogram of employee salary





```python
# create a scatter plot of the "Age" and "Employee Salary" features

plt.figure(figsize = (10, 5))
sns.scatterplot(data = data, x = 'age', y = 'salary', color = 'green', edgecolor = 'linen', alpha = 0.5)
plt.title('Scatterplot of Employee Age v/s Employee Salary')
plt.xlabel('Employee Age')
plt.ylabel('Employee Salary')
plt.show()
```

```python
# Create scatter plot of the "healthy_eating" and "active_lifestyle" features

plt.figure(figsize = (10,5))
sns.scatterplot(data = data, x = "healthy_eating", y = "active_lifestyle", color = 'red', edgecolor = 'linen', alpha = 0.5)
plt.title('Scatterplot of Healthy Eating v/s Active Lifestyle')
plt.xlabel('Healthy Eating')
plt.ylabel('Active Lifestyle')
plt.show()
```

```python
# Create a scatter plot of the "healthy_eating" and "salary" features

plt.figure(figsize = (10, 5))
sns.scatterplot(data = data, x = 'healthy_eating', y = 'salary', color = 'k', edgecolor = 'linen', alpha = 0.5)
plt.title('Scatterplot of Healthy Eating v/s Salary')
plt.xlabel('Healthy Eating')
plt.ylabel('Salary')
plt.show()
```

```python
# Create a count plot of the 'groups' feature

plt.figure(figsize = (10, 5))
sns.countplot(data = data, x = 'groups', edgecolor = 'linen', alpha = 0.5)
plt.title('Count Plot of Groups')
plt.xlabel('Groups')
plt.ylabel('Count')
plt.show()
```

```python
# Create a histogram of the 'Salary' features

plt.figure(figsize = (10, 5))
sns.histplot(data = data, x = 'salary', color = 'orange', edgecolor = 'linen', alpha = 0.7, bins = 10)
plt.title('Histogram of Salary')
plt.xlabel('Salary')
plt.ylabel('Count')
plt.show()
```
